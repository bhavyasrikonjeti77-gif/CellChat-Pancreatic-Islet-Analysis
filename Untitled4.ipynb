```python
%%R
library(Matrix)

# Identify gene columns
gene_cols <- setdiff(
  colnames(data_list[[1]]),
  c("barcode", "assigned_cluster")
)

# Check that all samples have the same genes
sapply(data_list, function(x) {
  identical(gene_cols, setdiff(
    colnames(x),
    c("barcode", "assigned_cluster")
  ))
})
```

```python
%%R
# Combine cell metadata
metadata <- do.call(rbind, lapply(seq_along(data_list), function(i) {
  data.frame(
    cell = paste0("Human_", i, "_", data_list[[i]]$barcode),
    cell_type = data_list[[i]]$assigned_cluster,
    sample = paste0("Human_", i)
  )
}))

# Extract expression data and convert to sparse matrices
expr_list <- lapply(data_list, function(x) {
  mat <- as.matrix(x[, gene_cols, drop = FALSE])
  mat <- Matrix(mat, sparse = TRUE)
  t(mat)
})

# Combine all samples
expr_matrix <- do.call(cbind, expr_list)

# Assign unique cell names
colnames(expr_matrix) <- metadata$cell
rownames(expr_matrix) <- gene_cols

# Check results
dim(expr_matrix)
head(metadata)
```

```python
%%R
# Check dimensions
dim(expr_matrix)

# Check number of cells in each sample
table(metadata$sample)

# Check different cell types
table(metadata$cell_type)

# Check for duplicate cell names
anyDuplicated(colnames(expr_matrix))

# Check metadata matching
identical(colnames(expr_matrix), metadata$cell)
```

```python
%%R
cat("Expression matrix dimensions:\n")
print(dim(expr_matrix))

cat("\nCells per sample:\n")
print(table(metadata$sample))

cat("\nCell types:\n")
print(table(metadata$cell_type))

cat("\nDuplicate cell names:\n")
print(anyDuplicated(colnames(expr_matrix)))

cat("\nMetadata matching:\n")
print(identical(colnames(expr_matrix), metadata$cell))
```

```python
%%R
# Identify duplicate cell names
duplicate_cells <- metadata$cell[duplicated(metadata$cell)]

cat("Total duplicate entries:", length(duplicate_cells), "\n")

# Display duplicate cell records
duplicate_records <- metadata[
  metadata$cell %in% duplicate_cells, ]

head(duplicate_records, 20)

# Check duplicates in each sample
for (s in unique(metadata$sample)) {
  cat("\n", s, ": ")
  print(sum(duplicated(metadata$cell[metadata$sample == s])))
}
```

```python
%%R
# Find all duplicated cell records
dup_idx <- which(
  duplicated(metadata$cell) |
  duplicated(metadata$cell, fromLast = TRUE)
)

duplicate_records <- metadata[dup_idx, ]

# Display first 20 duplicate records
print(head(duplicate_records, 20))

# Check whether duplicated cells have identical expression
first_dup <- duplicate_records$cell[1]

idx <- which(metadata$cell == first_dup)

cat("\nDuplicate cell:", first_dup, "\n")
cat("Cell types:\n")
print(metadata$cell_type[idx])

cat("\nIdentical expression profiles:\n")
print(identical(
  as.numeric(expr_matrix[, idx[1]]),
  as.numeric(expr_matrix[, idx[2]])
))
```

```python
%%R
# Select the first duplicated cell
first_dup <- duplicate_records$cell[1]

# Extract its original barcode
barcode_value <- sub("^Human_1_", "", first_dup)

# Find both records in the original dataset
original_rows <- data_list[[1]][
  data_list[[1]]$barcode == barcode_value, ]

# Display their barcode and cell type
print(original_rows[, 1:5])

# Check whether the complete records are identical
print(duplicated(original_rows))
```

```python
%%R
# Rebuild metadata using original cell IDs
metadata <- do.call(rbind, lapply(seq_along(data_list), function(i) {
  x <- data_list[[i]]

  data.frame(
    cell = paste0("Human_", i, "_", rownames(x)),
    cell_type = x$assigned_cluster,
    sample = paste0("Human_", i)
  )
}))

# Rebuild expression matrices
expr_list <- lapply(data_list, function(x) {
  mat <- as.matrix(x[, gene_cols, drop = FALSE])
  mat <- Matrix(mat, sparse = TRUE)
  t(mat)
})

# Combine all samples
expr_matrix <- do.call(cbind, expr_list)

# Assign corrected cell names
colnames(expr_matrix) <- metadata$cell
rownames(expr_matrix) <- gene_cols

# Verify
cat("Dimensions:\n")
print(dim(expr_matrix))

cat("\nDuplicate cell names:\n")
print(anyDuplicated(colnames(expr_matrix)))

cat("\nMetadata matching:\n")
print(identical(colnames(expr_matrix), metadata$cell))
```

```python
%%R

# Preserve the four human donor identities
metadata$samples <- factor(metadata$sample)

# Check sample distribution
print(table(metadata$samples))

# Recreate CellChat object
cellchat <- createCellChat(
  object = expr_matrix,
  meta = metadata,
  group.by = "cell_type"
)

# Assign human database
cellchat@DB <- CellChatDB.human

# Extract signaling-related genes
cellchat <- subsetData(cellchat)

# Check results
dim(cellchat@data.signaling)
```

```python
%%R

# Identify overexpressed genes using standard Wilcoxon test instead of presto
cellchat <- identifyOverExpressedGenes(cellchat, do.fast = FALSE)

# Identify overexpressed ligand-receptor interactions
cellchat <- identifyOverExpressedInteractions(cellchat)

# Check the results
cellchat
```

```python
%%R

# Calculate communication probabilities
cellchat <- computeCommunProb(
  cellchat,
  type = "triMean",
  population.size = TRUE
)

# Check communication results
dim(cellchat@net$prob)

# Display the first few communication probabilities
head(as.data.frame(cellchat@net$prob))
```
